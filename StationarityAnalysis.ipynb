```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.tsa.stattools import adfuller, kpss
```

```python
# Set plotting style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (15, 7)
```

```python
df_resampled = pd.read_csv('merged_data_1S.csv')
```

```python
print("Resampled DataFrame head:")
print(df_resampled.head())
print("Resampled DataFrame shape:", df_resampled.shape)
```

```python
# --- Stationarity Analysis ---
print("\n" + "="*50)
print("Stationarity Analysis of GPU Utilization")
print("="*50)

# Significance level
alpha = 0.05

# Get unique job-GPU combinations
job_gpu_combinations = df_resampled[['id_job', 'gpu_index']].drop_duplicates()

# Select a few sample combinations to demonstrate
# You might want to iterate through all or a representative subset for your thesis
sample_combinations = job_gpu_combinations.sample(min(5, len(job_gpu_combinations)), random_state=42) # Sample up to 5 unique jobs/GPUs

print(f"\nPerforming stationarity tests on {len(sample_combinations)} sample job-GPU combinations:\n")

for idx, row in sample_combinations.iterrows():
    job_id = row['id_job']
    gpu_idx = row['gpu_index']
    
    # Filter data for the specific job-GPU combination
    ts = df_resampled[(df_resampled['id_job'] == job_id) & 
                      (df_resampled['gpu_index'] == gpu_idx)]['utilization_gpu_pct_mean'].dropna()
    
    if len(ts) < 20: # ADF/KPSS need a reasonable number of observations
        print(f"Skipping Job {job_id}, GPU {gpu_idx} due to insufficient data points ({len(ts)}).")
        continue

    print(f"\n--- Analyzing Job: {job_id}, GPU: {gpu_idx} (Length: {len(ts)} points) ---")

    # Plot the time series
    plt.figure(figsize=(15, 4))
    plt.plot(ts.index, ts.values, label=f'Job {job_id} GPU {gpu_idx} Utilization')
    plt.title(f'GPU Utilization for Job {job_id}, GPU {gpu_idx}')
    plt.xlabel('Time (seconds since start of job segment)')
    plt.ylabel('Utilization (%)')
    plt.legend()
    plt.grid(True)
    plt.show()

    # Augmented Dickey-Fuller Test
    print("Augmented Dickey-Fuller Test:")
    try:
        adf_result = adfuller(ts)
        print(f"  ADF Statistic: {adf_result[0]:.4f}")
        print(f"  p-value: {adf_result[1]:.4f}")
        print("  Critical Values:")
        for key, value in adf_result[4].items():
            print(f"    {key}: {value:.4f}")
        
        if adf_result[1] <= alpha:
            print(f"  Result: p-value ({adf_result[1]:.4f}) <= {alpha}. Reject H0. The series IS stationary.")
        else:
            print(f"  Result: p-value ({adf_result[1]:.4f}) > {alpha}. Fail to reject H0. The series IS NOT stationary.")
    except ValueError as e:
        print(f"  ADF Test Error: {e}")
        print("  This can happen if the series is too short or has no variance.")


    # Kwiatkowski-Phillips-Schmidt-Shin (KPSS) Test
    print("\nKPSS Test (Level Stationarity):")
    try:
        kpss_result = kpss(ts, regression='c') # 'c' for constant mean, 'ct' for constant trend
        print(f"  KPSS Statistic: {kpss_result[0]:.4f}")
        print(f"  p-value: {kpss_result[1]:.4f}")
        print("  Critical Values:")
        for key, value in kpss_result[3].items():
            print(f"    {key}: {value:.4f}")

        if kpss_result[1] <= alpha:
            print(f"  Result: p-value ({kpss_result[1]:.4f}) <= {alpha}. Reject H0. The series IS NOT stationary.")
        else:
            print(f"  Result: p-value ({kpss_result[1]:.4f}) > {alpha}. Fail to reject H0. The series IS stationary.")
    except ValueError as e:
        print(f"  KPSS Test Error: {e}")
        print("  This can happen if the series is too short or has no variance.")

    print("\n" + "-"*60)

print("\nInterpretation Guide:")
print("---------------------")
print("Scenario 1: ADF (reject H0), KPSS (fail to reject H0)")
print("  - Both tests suggest stationarity. This is the ideal case for many traditional models.")
print("\nScenario 2: ADF (fail to reject H0), KPSS (reject H0)")
print("  - Both tests suggest non-stationarity. This is a clear indication the series needs differencing or other transformations.")
print("\nScenario 3: ADF (reject H0), KPSS (reject H0)")
print("  - Contradictory results. Series might be 'trend-stationary'. ADF says no unit root, but KPSS says not stationary around a constant mean/trend. This often means it's stationary around a mean but not around a trend, or vice-versa, depending on the KPSS `regression` parameter. Consider removing a trend first, then re-testing.")
print("\nScenario 4: ADF (fail to reject H0), KPSS (fail to reject H0)")
print("  - Contradictory results. Series might be 'difference-stationary' but not 'trend-stationary'. ADF suggests non-stationarity, but KPSS suggests stationarity. This can happen if the data is white noise and a unit root test fails to detect it.")
print("\nGeneral Advice:")
print(" - If a series is non-stationary, common techniques like differencing (e.g., `df['col'].diff().dropna()`) can be applied to make it stationary before modeling with traditional methods like ARIMA.")
print(" - For models like LSTMs or XGBoost with engineered features (lags, rolling stats), strict stationarity is often less of a direct requirement, as these models can learn and capture non-stationary patterns. However, understanding stationarity still provides insights into the data's underlying dynamics.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# --- Run the tests as before ---
import pandas as pd
import numpy as np
from statsmodels.tsa.stattools import adfuller, kpss
from tqdm import tqdm

def adf_test(series, alpha=0.05):
    try:
        result = adfuller(series.dropna(), autolag='AIC')
        return result[1] < alpha  # p < alpha = stationary
    except Exception:
        return False

def kpss_test(series, alpha=0.05):
    try:
        result = kpss(series.dropna(), regression='c', nlags="auto")
        return result[1] > alpha  # p > alpha = stationary
    except Exception:
        return False

results = []
for job_id, group in tqdm(df.groupby("id_job")):
    series = group.sort_values("timestamp")["utilization_gpu_pct"]

    adf_res = adf_test(series)
    kpss_res = kpss_test(series)

    results.append({
        "job_id": job_id,
        "adf_stationary": adf_res,
        "kpss_stationary": kpss_res,
        "tests_passed": int(adf_res) + int(kpss_res)
    })

stationarity_df = pd.DataFrame(results)

# Count jobs by category (0,1,2 tests passed)
summary_counts = stationarity_df["tests_passed"].value_counts().sort_index()
print(summary_counts)

# --- Select representative jobs for visualization ---
job_0 = stationarity_df.query("tests_passed == 0").sample(1, random_state=42)["job_id"].iloc[0]
job_1 = stationarity_df.query("tests_passed == 1").sample(1, random_state=42)["job_id"].iloc[0]
job_2 = stationarity_df.query("tests_passed == 2").sample(1, random_state=42)["job_id"].iloc[0]

example_jobs = [job_0, job_1, job_2]

# --- Plot GPU utilization for each example job ---
fig, axes = plt.subplots(3, 1, figsize=(12, 10), sharex=False)

for i, job in enumerate(example_jobs):
    job_series = df[df["id_job"] == job].sort_values("timestamp")["utilization_gpu_pct"].reset_index(drop=True)
    
    sns.lineplot(data=job_series, ax=axes[i])
    axes[i].set_title(f"Job {job} - Tests Passed: {stationarity_df.loc[stationarity_df.job_id==job, 'tests_passed'].values[0]}")
    axes[i].set_ylabel("GPU Utilization (%)")
    axes[i].set_xlabel("Time step")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
from statsmodels.tsa.stattools import adfuller, kpss
from tqdm import tqdm

# --- Helper functions ---
def adf_test(series, alpha=0.05):
    try:
        result = adfuller(series.dropna(), autolag='AIC')
        return result[1] < alpha  # p < alpha = stationary
    except Exception:
        return False

def kpss_test(series, alpha=0.05):
    try:
        result = kpss(series.dropna(), regression='c', nlags="auto")
        return result[1] > alpha  # p > alpha = stationary
    except Exception:
        return False

# --- Apply to each (job_id, gpu_index) ---
results = []
for (job_id, gpu), group in tqdm(df.groupby(["id_job", "gpu_index"])):
    group = group.sort_values("timestamp")

    util = group["utilization_gpu_pct"]

    adf_res = adf_test(util)
    kpss_res = kpss_test(util)

    # classify the result more explicitly
    if adf_res and kpss_res:
        category = "Both passed"
    elif adf_res and not kpss_res:
        category = "ADF only"
    elif not adf_res and kpss_res:
        category = "KPSS only"
    else:
        category = "Neither"

    results.append({
        "job_id": job_id,
        "gpu_index": gpu,
        "adf_stationary": adf_res,
        "kpss_stationary": kpss_res,
        "category": category
    })

stationarity_df = pd.DataFrame(results)

# --- Summaries ---
print("Counts by category:")
print(stationarity_df["category"].value_counts())

print("\nPreview:")
print(stationarity_df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from scipy.signal import detrend

# ----------------------------
# Helper: normalize + differencing with seconds axis
# ----------------------------
def plot_differencing(series_df, job_id, gpu_idx, minutes=20):
    # Normalize
    scaler = MinMaxScaler()
    series_df["util_norm"] = scaler.fit_transform(series_df[["utilization_gpu_pct"]])

    # Apply differencing
    series_df["util_diff"] = series_df["util_norm"].diff()

    # Create seconds axis relative to first timestamp
    t0 = series_df["timestamp"].min()
    series_df["seconds"] = (series_df["timestamp"] - t0).dt.total_seconds()

    # Take first N minutes
    cutoff = minutes * 60
    subset = series_df[series_df["seconds"] <= cutoff]

    # --- Plot ---
    fig, axes = plt.subplots(2, 1, figsize=(12, 6), sharex=True)
    sns.lineplot(x=subset["seconds"], y=subset["util_norm"], ax=axes[0])
    axes[0].set_title(f"Job {job_id}, GPU {gpu_idx} - Normalized Utilization (First {minutes} min)")
    axes[0].set_ylabel("Normalized Utilization [0,1]")

    sns.lineplot(x=subset["seconds"], y=subset["util_diff"], ax=axes[1], color="orange")
    axes[1].set_title("After First Differencing")
    axes[1].set_xlabel("Time (seconds)")
    axes[1].set_ylabel("Differenced Utilization")

    plt.tight_layout()
    plt.show()


# ----------------------------
# Helper: detrending with seconds axis
# ----------------------------
def plot_detrending(series_df, job_id, gpu_idx, minutes=20):
    # Normalize
    scaler = MinMaxScaler()
    series_df["util_norm"] = scaler.fit_transform(series_df[["utilization_gpu_pct"]])

    # Apply detrending (linear)
    series_df["util_detrend"] = detrend(series_df["util_norm"])

    # Create seconds axis
    t0 = series_df["timestamp"].min()
    series_df["seconds"] = (series_df["timestamp"] - t0).dt.total_seconds()

    # Take first N minutes
    cutoff = minutes * 60
    subset = series_df[series_df["seconds"] <= cutoff]

    # --- Plot ---
    fig, axes = plt.subplots(2, 1, figsize=(12, 6), sharex=True)
    sns.lineplot(x=subset["seconds"], y=subset["util_norm"], ax=axes[0])
    axes[0].set_title(f"Job {job_id}, GPU {gpu_idx} - Normalized Utilization (First {minutes} min)")
    axes[0].set_ylabel("Normalized Utilization [0,1]")

    sns.lineplot(x=subset["seconds"], y=subset["util_detrend"], ax=axes[1], color="green")
    axes[1].set_title("After Detrending (Linear)")
    axes[1].set_xlabel("Time (seconds)")
    axes[1].set_ylabel("Detrended Utilization")

    plt.tight_layout()
    plt.show()


# ----------------------------
# Run examples
# ----------------------------

# Pick one series "ADF only"
adf_only = stationarity_df.query("category == 'ADF only'").iloc[0]
job_id, gpu_idx = adf_only["job_id"], adf_only["gpu_index"]
series_df = (
    df[(df["id_job"] == job_id) & (df["gpu_index"] == gpu_idx)]
    .sort_values("timestamp")[["timestamp", "utilization_gpu_pct"]]
    .reset_index(drop=True)
)
series_df["timestamp"] = pd.to_datetime(series_df["timestamp"])
plot_differencing(series_df.copy(), job_id, gpu_idx, minutes=40)


# Pick one series "KPSS only"
non_stat = stationarity_df.query("category == 'KPSS only'").iloc[0]
job_id, gpu_idx = non_stat["job_id"], non_stat["gpu_index"]
series_df = (
    df[(df["id_job"] == job_id) & (df["gpu_index"] == gpu_idx)]
    .sort_values("timestamp")[["timestamp", "utilization_gpu_pct"]]
    .reset_index(drop=True)
)
series_df["timestamp"] = pd.to_datetime(series_df["timestamp"])
plot_detrending(series_df.copy(), job_id, gpu_idx, minutes=40)
```
